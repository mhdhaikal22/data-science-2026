```python
import pandas as pd
import numpy as np

# Muat Dataset Titanic
URL = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(URL)

# Q1. Berapa jumlah baris dan kolom dataset Titanic?

print("Jumlah baris:", df.shape[0])
print("Jumlah kolom:", df.shape[1])
```

```python
# Q2. Kolom mana saja yang memiliki missing values? Berapa jumlah dan proporsinya (%) per kolom?

missing = df.isnull().sum()
missing = missing[missing > 0]

for kolom in missing.index:
    print(
        f"{kolom}: {missing[kolom]} missing values "
        f"({missing[kolom] / len(df) * 100:.2f}%)"
    )
```

```python
# Q3. Berapa persen penumpang yang selamat (Survived=1)? Gunakan operasi NumPy atau Pandas.

persen_selamat = df['Survived'].mean() * 100

print(f"Persentase penumpang yang selamat: {persen_selamat:.2f}%")
print("Perhitungan menggunakan operasi Pandas")
```

```python
# Q4. Filter penumpang wanita dari kelas 1 (Pclass=1). Berapa jumlahnya dan berapa rata-rata usianya?

wanita_kelas1 = df[
    (df['Sex'] == 'female') &
    (df['Pclass'] == 1)
]

print(f"Jumlah penumpang wanita kelas 1: {len(wanita_kelas1)}")
print(f"Rata-rata usia penumpang wanita kelas 1: {wanita_kelas1['Age'].mean():.2f}")
```

```python
# Q5. Gunakan groupby untuk menghitung tingkat keselamatan (mean Survived) per kelas penumpang. Kelas mana yang paling tinggi?

tingkat_keselamatan = df.groupby('Pclass')['Survived'].mean() * 100

print("Tingkat keselamatan per kelas (%):")
print(tingkat_keselamatan.round(2))

kelas_tertinggi = tingkat_keselamatan.idxmax()

print(f"\nKelas dengan tingkat keselamatan tertinggi: Kelas {kelas_tertinggi}")
```

```python
# Q6. Buat analisis tambahan berdasarkan variabel lain yang Anda anggap menarik. Sertakan interpretasi singkat.

print("Saya memilih variabel jenis kelamin (Sex) untuk dianalisis.")

survival_by_gender = df.groupby('Sex')['Survived'].mean() * 100

print("\nTingkat keselamatan berdasarkan jenis kelamin (%):")
print(survival_by_gender.round(2))

print("\nInterpretasi:")
print("Penumpang wanita memiliki tingkat keselamatan yang jauh lebih tinggi (74.20%)")
print("dibandingkan penumpang pria (18.89%). Hal ini menunjukkan bahwa dalam proses")
print("evakuasi Titanic, wanita memiliki prioritas yang lebih besar untuk diselamatkan.")
```
