```python
# Load Dataset

import pandas as pd
import numpy as np
from scipy.stats.mstats import winsorize

df = pd.read_csv("https://drive.google.com/uc?id=1LfQWProB0VjWN5q8bKuRIgn-stULfIRo")
```

```python
# Eksplorasi Awal

print("Informasi Dataset:")
df.info()

print("\nStatistik Deskriptif:")
print(df.describe())

print("\nJumlah Missing Values:")
print(df.isnull().sum())
```

```python
# Hapus Duplikat

print("Shape sebelum hapus duplikat:", df.shape)

df.drop_duplicates(inplace=True)

print("Shape setelah hapus duplikat:", df.shape)
```

```python
# Normalisasi String

df['kota'] = df['kota'].str.strip().str.title()
df['kondisi'] = df['kondisi'].str.strip().str.lower()

print("Normalisasi string selesai.")

print(df['kota'].unique())
print(df['kondisi'].unique())
```

```python
# Imputasi Missing Values

print("Jumlah missing values sebelum imputasi:")
print(df[['luas_m2', 'harga_juta', 'kamar']].isnull().sum())

df['luas_m2'] = df['luas_m2'].fillna(df['luas_m2'].median())
df['harga_juta'] = df['harga_juta'].fillna(df['harga_juta'].median())
df['kamar'] = df['kamar'].fillna(df['kamar'].mode()[0])

print("\nJumlah missing values setelah imputasi:")
print(df[['luas_m2', 'harga_juta', 'kamar']].isnull().sum())
```

```python
# Tangani Outlier (IQR Fence)

for col in ['harga_juta', 'luas_m2', 'tahun_bangun']:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    print(f"\nKolom: {col}")
    print(f"Batas bawah: {lower:.2f}")
    print(f"Batas atas : {upper:.2f}")

    df[col] = df[col].clip(lower=lower, upper=upper)

print("\nPenanganan outlier selesai menggunakan metode IQR Fence.")
```

```python
# Validasi Akhir

print("Total missing values:", df.isnull().sum().sum())
print("Total data duplikat:", df.duplicated().sum())

assert df.isnull().sum().sum() == 0, "Masih ada missing values!"
assert df.duplicated().sum() == 0, "Masih ada data duplikat!"

print("Validasi berhasil. Dataset sudah bersih.")
```

```python
# Ekspor Dataset Bersih

df.to_csv('housing_clean.csv', index=False)

print("Dataset bersih berhasil disimpan sebagai housing_clean.csv")
```
